## Solved dataset Using Pandas and Matplotlib


```python
import pandas as pd
```

```python
path = pd.read_excel(r"C:\Users\91702\Downloads\flipkart_sales_data.xlsx")
df  = pd.DataFrame(path)
df
```

```python
df.info()
```

```python
df.dropna(inplace = True)
```

```python
df.info()
```

```python
df.describe()
```

```python
s = df["Sales Amount"].sum()
print("Total sales:", f'{s:,.0f}')
```

```python
df
```

```python
df["Month_Name"] = df["Order Date"].dt.month_name()
df
```

```python
df["Weekday Name"]= df["Order Date"].dt.day_name()
df
```

```python
df.groupby("Product Category")["Sales Amount"].sum()
```

```python
import matplotlib.pyplot as plt
ct = df.groupby("Product Category")["Sales Amount"].sum()

plt.figure(figsize=(10,4))
b = plt.bar(ct.index, ct.values, color  ="gold",alpha = 0.5,edgecolor= "maroon")

for bar in b:
    plt.text(bar.get_x()+bar.get_width()/2,bar.get_height(),f'{bar.get_height():,.0f}',va= "bottom",ha ="center",fontsize= 8)


plt.xlabel("Product Category",fontsize = 10)
plt.ylabel("Total Sales" , fontsize = 10)
plt.title("Total Sales by Product Category")

plt.tight_layout()
plt.show()
```

```python
df
```

```python
p = df.groupby("Payment Method")["Order ID"].count()

plt.figure(figsize= (10,5))
plt.pie(p,colors = ("lightgreen","gold","skyblue","lightpink","brown","purple"),labels= p.index,autopct = "%0.2f%%")

plt.title("Payment mode")
plt.show()
```

```python
df
```

```python
df["Year"] = df["Order Date"].dt.year
df
```

```python
m = df.groupby("Month_Name")["Sales Amount"].sum()

plt.figure(figsize = (15,4))
plt.plot(m.index,m.values,color= "red",alpha = 0.5,linestyle = "-",marker = "o")
plt.fill_between(m.index,m.values,color= "gold",alpha =0.3)

for x,y in zip(m.index,m.values):
    plt.text(x,y,f'{y:,.0f}', va = "bottom", ha = "left", fontsize = 6)

plt.xlabel("Month")
plt.ylabel("Total sales")
plt.title("Month wise Total Sales")

plt.show()
```

```python
w = df.groupby("Weekday Name")["Sales Amount"].sum()

plt.figure(figsize = (15,5))
plt.plot(w.index,w.values,marker = "o",linestyle = "--")

for x,y in zip(w.index,w.values):
    plt.text(x,y,f'{y:,.0f}',va = "top", ha ="left", fontsize = 8)

plt.title("Weekday wise Total sales")
plt.xlabel("Weekday",fontsize =10)
plt.ylabel("Total sales",fontsize = 10)

plt.show()
```
