## DataFrames


```python
# DataFrames contains mutiple columns and rows
# Creating DataFrames

import pandas as pd
data = {"Name" : ["Rakshitha","Rajesh","Ramya","Bhavya"],
       "Salary" : [65000,45000,68000,53000],
       "Department" : ["IT","SALES","IT","HR"]}
df = pd.DataFrame(data)
df
```

```python
# To obtain Top n and Bottom n values
print(df.head(1))
print("\n")
print(df.tail(1))
```

```python
# Sort the values based on salary column from dataframe
df["Salary"].sort_values(ascending = True)
```

```python
## Sort salary column in descending order
df["Salary"].sort_values(ascending = False)
```

```python
# To obtain Informations about dataset
df.info()
```

```python
# To Obtain statistical values of SALARY column in dataset
df.describe()
```

```python
# Let us consider an example :
# An additional column called BONUS SALARY has to be added to the dataset : Bonus is of 5000

df["Bonus Salary"] = df["Salary"]+5000

# Lets create another column that specifies the Age of Employees

df["Age"]=[23,26,43,56]

# Read the data
df
```

```python
# To rename any Columns : Lets Drop age column
df.drop("Age",axis = 1, inplace = True)

# INPLACE is used to permanently delete columns from dataset 

# Read the data
df
```

```python
# To rename any column name : let us consider Bonus Salary has to converted into B_salary

df.rename(columns = {"Bonus Salary": "B_salary"},inplace =  True)

df
```
