## Pandas Basics

```python
## In Pandas their are two types of data Structures
## SERIES AND DATAFRAME
```

```python
## Series: Contains Single column and multiples rows
## DataFrame: Contain proper structure with multiple column and rows
```

```python
## To craete data Base
# import pandas

import pandas as pd

# Create dict which contains column name and values
data = ["Rakshan","Ramya","Rajesh","Bhavya"]

# Create into series 
series = pd.Series(data)
series 
```

```python
## To Obtain information about datatypes 
series.info()
```

```python
# To obtain top 1 data
series.head(1)
```

```python
# To fetch Bottom 1 data
series.tail(1)
```

```python
## If we want to replace index value with a...b...c...d or others
index = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
```

```python
index
```

```python
# To sort values in Ascending order
# If (ascending = True) is No specified by default it will sort in ASCENDING ORDER
index.sort_values()
```

```python
# To sort Values in descending order
index.sort_values(ascending = False)
```
