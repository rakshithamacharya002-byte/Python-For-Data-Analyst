```python
# To create pie chart :: plt.pie
# Lets consider a dataset

import pandas as pd
import matplotlib.pyplot as plt
path = pd.read_csv(r"C:\Users\91702\Downloads\blinkit_data.csv")

df = pd.DataFrame(path)
df
```

```python
# From above dataset lets clean the data and obtain the all the additional colums which are vrequired for us ::
df["Item Fat Content"].unique()

## Replace the values
df.replace(["LF","reg","low fat"],["Low Fat","Regular","Low Fat"],inplace = True)
```

```python
df.info()
```

```python
# TO FIND TOTAL SALES BY TYPE OF ITEM
# We have make us e to group by and aggregate functions

Sales_by_item_type = df.groupby("Item Type")["Sales"].sum().sort_values()
Sales_by_item_type
# We can also sort the items by total sales :: df.groupby("Item Type")["Sales"].sum().sort_values()
# If descending order :: df.groupby("Item Type")["Sales"].sum().sort_values(ascending = True)

# We have Index and values in below OUTPUT
```

```python
# Lets create bar chart for above Grouped total sales by item type
# For these type of Real world data we have consider :: index ---- x cordinate and values ----- y cordinate

# Lets provide color,border and along with we can also fix charts width and height :: plt.figure(figsize = (width_size,height_size))
plt.figure(figsize = (25,3))
plt.bar(Sales_by_item_type.index,Sales_by_item_type.values)

plt.show()
```

```python
# Lets create gradient shading of color by using normalize:: Lets fetch max and min from Sales_by_item_type---values
# NORMALIZE SYNTAX :: plt.Normalize(min(value),max(value))
# After fetching Max and min create a CM :: color mapping
# SYNTAX :: plt.cm.Blues(norm(values))

norm = plt.Normalize(min(Sales_by_item_type.values),max(Sales_by_item_type.values))
clr = plt.cm.Blues(norm(Sales_by_item_type.values))

plt.figure(figsize = (25,3))
plt.bar(Sales_by_item_type.index,Sales_by_item_type.values,color = clr )

for x,y in zip(Sales_by_item_type.index, Sales_by_item_type.values):
    plt.text(x,y,f'{y:,.0f}',va = "bottom",ha ="center",fontsize = 10)

plt.show()
```

```python
# Lets create Pie chart :: plt.pie(Grouped_value _name, Lables = index ,autopct)
# To get values in Pie chart Use :: autopct = "%0.1f%%" for 1 decimal point
#                                   autopct = "%0.0f%%" for 0 decimal point

Size_total_sales = df.groupby("Outlet Size")["Sales"].sum()
plt.pie(Size_total_sales,labels = Size_total_sales.index,autopct = "%0.0f%%")

plt.show()
```

```python
# Lets CREATE AREA CHART : Similar to line chart but in area chart we have to shade the areas of index and values
# plt.fill_between(index,values)

# Lets create Yearly sum of sales

year_sales = df.groupby("Outlet Establishment Year")["Sales"].sum()
plt.plot(year_sales.index,year_sales.values)
plt.fill_between(year_sales.index,year_sales.values,alpha = 0.2)

```
